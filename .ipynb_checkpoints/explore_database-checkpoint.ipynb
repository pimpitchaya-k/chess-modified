```python
import sqlite3
import pickle
```

```python
db_path = "./data/dev/dev_databases/preprocessed/context_vector_db/chroma.sqlite3"

# Connect to the SQLite database
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

# List all tables inside the database
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tables = cursor.fetchall()
print("Tables:", [table[0] for table in tables])

# To query a specific table (replace 'table_name' with a real table from the list above)
# cursor.execute("SELECT * FROM table_name LIMIT 5;")
# print(cursor.fetchall())

conn.close()
```

```python
file_path = "./data/dev/dev_databases/card_games/preprocessed/card_games_lsh.pkl"

with open(file_path, 'rb') as f:
    lsh = pickle.load(f)

print(type(lsh))
```

```python
lsh
```

```python
# Check basic LSH parameters
print("Number of hash permutations (h):", lsh.h)
print("Number of bands (b):", lsh.b)
print("Number of rows per band (r):", lsh.r)

print("\nIs LSH empty?:", lsh.is_empty())

# View the stored keys (the IDs/names of the items indexed in the database)
if hasattr(lsh, 'keys'):
    # Depending on the datasketch version, keys might be a dict or a set-like object
    # For datasketch 1.5+, lsh.keys is a dict mapping keys to their hash buffers
    keys_list = list(lsh.keys.keys()) if isinstance(lsh.keys, dict) else list(lsh.keys)
    
    print(f"\nTotal items indexed: {len(keys_list)}")
    print("First 10 items:", keys_list[:10])

# See all internal attributes and methods available to you
print("\nAvailable methods/attributes:", [m for m in dir(lsh) if not m.startswith('_')])
```
