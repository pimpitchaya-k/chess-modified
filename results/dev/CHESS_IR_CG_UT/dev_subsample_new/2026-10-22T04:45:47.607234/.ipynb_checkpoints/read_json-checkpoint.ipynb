```python
import json
from pathlib import Path

path = Path("/data/home/mayppcy/patient_case_search/master_thesis/baseline_models/CHESS/results/dev/CHESS_IR_CG_UT/dev_subsample_200/2026-10-22T04:45:47.607234")
dont_execute = ['-args', '-predictions', '-statistics']

results = {}
exec_res = []
for p in sorted(path.glob("*.json")):
    if any(name in p.stem for name in dont_execute):
        continue
    with open(p, "r", encoding="utf-8") as f:
        results[p.stem] = json.load(f)
        if results[p.stem][-1]['tool_name']=='execution_accuracy':
            exec_res.append(results[p.stem][-1]['final_SQL']['exec_res'])
print(len(results))
```

```python
print(exec_res)
```

```python
n = len(exec_res)
```

```python
cleaned_exec_res = [0 if i=='error' else i for i in exec_res]
print(cleaned_exec_res)
```

```python
sum(cleaned_exec_res)/n
```
